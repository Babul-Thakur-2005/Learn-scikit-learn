# LOADING DATASET

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris


iris = load_iris()

x = iris.data
y = iris.target

feature_names = iris.feature_names
target_names = iris.target_names


print("Features names - ", feature_names)
print("target name - ", target_names)

print("\n Type of X is :", type(x))
print("\n type of Y is :", type(y))


print("\nFirst 5 rows of X :\n", x[:5])
```

# SPLITING DATASET

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.4, random_state = 1)
```

```python
print("x_train shape - ", x_train.shape)
print("x_test shape - ", x_test.shape)
print("y_train shape - ", y_train.shape)
print("y_test shape - ", y_test.shape)
```

# HANDLING CATEGORICAL DATASET

# 1.LABEL ENCODING

```python
# LabelEncoder() -> encoder object that convert categorical object into numerical labels.
# fit_transform() -> firstly fits the categorical data and then convert into corresponding numerical labels.
```

```python
from sklearn.preprocessing import LabelEncoder

categorical_feature  = ["cat", "dog", "dog", "cat", "bird"]

encoder = LabelEncoder()

encoded_feature = encoder.fit_transform(categorical_feature)

print("Encoded feature :", encoded_feature)
```

# 2.ONE HOT ENCODING

```python
from sklearn.preprocessing import OneHotEncoder

categorical_feature = ['cat', "dog", 'dog', 'cat', "bird"]

categorical_feature = np.array(categorical_feature).reshape(-1,1)

encoder = OneHotEncoder(sparse_output = False)

encoded_feature = encoder.fit_transform(categorical_feature)

print("OneHotEncoded Feature: \n", encoded_feature)
```

# TRAINING THE MODEL

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter = 200)

log_reg.fit(x_train, y_train)
```

# MAKE PREDICTIONS

```python
y_pred = log_reg.predict(x_test)
```

# EVALUATING MODEL ACCURACY

```python
from sklearn import metrics

print("Logistic Regression model accuracy:", metrics.accuracy_score(y_test, y_pred))
```
